```python
# Exercise 4
def func4(a, b):
    b.sort() # Time complexity for this function is O(Nlog N) where N = len(b)
    intersect = 0
    for x in a:
        if binary_search(b, x) >= 0:
            intersect = intersect + 1
    return intersect

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
# Sample function call: func4([1, 2, 3, 4], [2, 4, 6, 8])
print(func4([1, 2, 3, 4], [2, 4, 6, 8]))
```

```python
# Exercise 5
def func5(n):
    guess = 1
    while guess * guess <= n:
        if guess * guess == n:
            return guess
        guess = guess + 1
    return None
```

```python
# Sample function call: func5(16)
print(func5(16))
```

```python
# Exercise 6
def func8(A):
    func8_helper(A, 0, len(A) - 1)
    return A

def func8_helper(A, i, j):
    if i >= j:
        return

    # If first element is larger than last, swap them
    if A[i] > A[j]:
        A[i], A[j] = A[j], A[i]

    # If there are more than 2 elements in the array
    if j - i + 1 > 2:
        m = (j - i + 1) // 3
        func8_helper(A, i, j - m)
        func8_helper(A, i + m, j)
        func8_helper(A, i, j - m)
```

```python
#Sample function call: func8([3,1,2,7,8,6])
print(func8([3, 1, 2, 7, 8, 6]))
```

```python
# Exercise 7
def func12(arr, k):
    arr_len = len(arr)
    k = k % arr_len
    # Initialize a new array of the same length as 'arr'
    custom_arr = [0] * arr_len

    for i in range(arr_len):
        mystery_index = (i + k) % arr_len
        custom_arr[mystery_index] = arr[i]
    return custom_arr
```

```python
#Sample function call: func12([1, 2, 3, 4, 5], 2)
print(func12([1, 2, 3, 4, 5], 2))
```

```python
# write a function named as sum_arr to find sum of elements in given array
def sum_arr(arr):
    total = 0
    for x in arr:
        total = total + x
    return total

arr = [12, 3, 4, 15, 45, 78, 450]
print(sum_arr(arr))
```

```python
#write a function named as max_min to find minimum (or maximum) element in an array
# 1. Define the function
def max_min(arr, find_max=True):
    if not arr:
        return "Array is empty"


    result = arr[0]

    for num in arr:
        if find_max:

            if num > result:
                result = num
        else:

            if num < result:
                result = num

    return result


my_list = [15, 82, 3, 45, 91, 12]

maximum = max_min(my_list, find_max=True)
minimum = max_min(my_list, find_max=False)


print(f"The Array: {my_list}")
print(f"The Maximum element is: {maximum}")
print(f"The Minimum element is: {minimum}")
```

```python
# write a function named as Counting_EvenOdd to count number of even and odd elements in an array
def Counting_EvenOdd(arr):
    even_count = 0
    odd_count = 0

    for num in arr:

        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

    return even_count, odd_count


arr = [12, 1234, 45, 67, 1]


evens, odds = Counting_EvenOdd(arr)

print(f"Array: {arr}")
print(f"Number of Even elements: {evens}")
print(f"Number of Odd elements: {odds}")
```

```python


# write a function named as linear_Search to perform a linear search in 1D-array
def linear_Search(arr, target):

    for i in range(len(arr)):
        if arr[i] == target:
            return i

    return -1


my_arr = [10, 23, 45, 70, 11, 15]
search_val = 70

result = linear_Search(my_arr, search_val)

if result != -1:
    print(f"Element {search_val} found at index: {result}")
else:
    print(f"Element {search_val} not found in the array.")
```

```python
# write a function named as Binary_Search to perform binary search in 1D-array
def Binary_Search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1
```

```python


# call the function Binary_Search for array arr = [1, 3, 5, 7, 9,15,16,14,45]


arr = [1, 3, 5, 7, 9, 15, 16, 14, 45]

arr.sort()
print("Sorted Array:", arr)

target = 14
result = Binary_Search(arr, target)

if result != -1:
    print("Element", target, "found at index:", result)
else:
    print("Element not found")
```
