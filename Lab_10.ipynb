# Warshall algorithm to find reachability between all pairs of vertices in a graph.
Problem Description:
Understand transitive closure of a graph. Use Python to compute reachability using matrix method.

#Theory
Warshall algorithm finds whether a path exists between every pair of vertices. It updates the adjacency matrix step by step. If there is a path from i to j through k, then the entry becomes 1.

## Algorithm
1. Represent graph using adjacency matrix.
2. For each intermediate vertex k
3. Check all pairs i and j
4. Update value if path exists through k
4. Repeat for all vertices


```python
def warshall(graph):
    n = len(graph)
    for k in range(n):
        for i in range(n):
            for j in range(n):
                graph[i][j] = graph[i][j] or (graph[i][k] and graph[k][j])
    return graph

def floyd(graph):
    n = len(graph)
    dist = [[graph[i][j] if graph[i][j] != 0 else float('inf') for j in range(n)] for i in range(n)]
    for i in range(n):
        dist[i][i] = 0
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]
    return dist

def print_matrix(matrix):
    for row in matrix:
        print(row)

def readable_paths(matrix):
    n = len(matrix)
    print("\nPath existence between vertices:")
    for i in range(n):
        for j in range(n):
            if i != j:
                if matrix[i][j] != 0 and matrix[i][j] != float('inf'):
                    print(f"Path exists from {i+1} to {j+1}")
                else:
                    print(f"No path from {i+1} to {j+1}")

# --- MAIN PROGRAM ---
n = 5
print(f"Enter adjacency matrix for {n} vertices (use 0/1):")
graph = []
for i in range(n):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    if len(row) != n:
        raise ValueError(f"Row {i+1} must have {n} entries, but got {len(row)}")
    graph.append(row)

print("\nOriginal Graph:")
print_matrix(graph)

# Warshall
warshall_result = warshall([row[:] for row in graph])
print("\nTransitive Closure (Warshall):")
print_matrix(warshall_result)
readable_paths(warshall_result)

# Floyd
floyd_result = floyd(graph)
print("\nShortest Path Matrix (Floyd):")
print_matrix(floyd_result)
readable_paths(floyd_result)
```

## Find shortest paths between all pairs of vertices using Floyd Warshall algorithm.

#Understand all pairs shortest path problem. Implement dynamic programming approach in Python.

* Given a weighted graph with positive or negative edge weights, find
shortest distances between every pair of vertices.

# Theory
Floyd Warshall algorithm computes shortest paths between all pairs of nodes. It updates the distance matrix by checking if a shorter path exists through an intermediate vertex. Works with negative weights but fails with negative cycles.

Algorithm
1. Represent graph using distance matrix.
2. Initialize matrix with given weights.
3. Set distance of node to itself as 0.
4. For each intermediate vertex k
5. Update distance[i][j] = min(distance[i][j], distance[i][k] + distance[k][j])
6. Repeat for all vertices

```python
def floyd_warshall(graph):

    n = len(graph)

    dist = []
    for i in range(n):
        dist.append(graph[i][:])

    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]

    return dist

```

```python
INF = float('inf')

graph = [
[0, 3, INF, 7],
[8, 0, 2, INF],
[5, INF, 0, 1],
[2, INF, INF, 0]
]

result = floyd_warshall(graph)

print("Shortest distance matrix:")

for row in result:
    print(row)
```

```python
INF = float('inf')

def floyd_warshall(graph):
    n = len(graph)
    dist = [[graph[i][j] for j in range(n)] for i in range(n)]
    # Floyd–Warshall core
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]
    return dist

def detect_negative_cycle(dist):
    n = len(dist)
    for i in range(n):
        if dist[i][i] < 0:
            return True
    return False

def dijkstra(graph, src):
    n = len(graph)
    dist = [INF] * n
    dist[src] = 0
    visited = [False] * n
    for _ in range(n):
        # pick min distance unvisited
        u = min((d, idx) for idx, d in enumerate(dist) if not visited[idx])[1]
        visited[u] = True
        for v in range(n):
            if graph[u][v] != INF and dist[u] + graph[u][v] < dist[v]:
                dist[v] = dist[u] + graph[u][v]
    return dist

#  MAIN PROGRAM
n = int(input("Enter number of vertices: "))
print(f"Enter adjacency matrix ({n}x{n}), use INF for no edge:")

graph = []
for i in range(n):
    row = input(f"Row {i+1}: ").split()
    # convert "INF" to float('inf')
    row = [INF if x.upper()=="INF" else int(x) for x in row]
    if len(row) != n:
        raise ValueError(f"Row {i+1} must have {n} entries")
    graph.append(row)

# Floyd–Warshall
dist_fw = floyd_warshall(graph)
print("\nShortest distance matrix (Floyd–Warshall):")
for row in dist_fw:
    print(row)

# Negative cycle detection
if detect_negative_cycle(dist_fw):
    print("\nGraph contains a negative cycle!")
else:
    print("\nNo negative cycle detected.")

# Shortest path between two vertices
src = int(input("Enter source vertex (1-based): ")) - 1
dst = int(input("Enter destination vertex (1-based): ")) - 1
if dist_fw[src][dst] == INF:
    print(f"No path exists from {src+1} to {dst+1}")
else:
    print(f"Shortest path from {src+1} to {dst+1} is {dist_fw[src][dst]}")

# Compare with Dijkstra
print("\nDijkstra results from source", src+1)
dist_dij = dijkstra(graph, src)
print(dist_dij)
```
